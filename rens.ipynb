```python
import pandas as pd

df = pd.read_csv("data/model_data.csv")
print(df.shape)

mangler = df.isna().sum()
mangler[mangler > 0]
```

```python
df["contract_missing"] = df["contract_years_left"].isna().astype(int)
median_kontrakt = df["contract_years_left"].median()
df["contract_years_left"] = df["contract_years_left"].fillna(median_kontrakt)

print("Median kontrakt:", round(median_kontrakt, 2), "år")
print("Spillere uten kontraktsdato:", df["contract_missing"].sum())
```

```python
df["height_in_cm"] = df["height_in_cm"].fillna(df["height_in_cm"].median())
df["foot"] = df["foot"].fillna("unknown")
df["international_caps"] = df["international_caps"].fillna(0)
df["international_goals"] = df["international_goals"].fillna(0)
```

```python
df["league"] = df["current_club_domestic_competition_id"].fillna("annet")
df["league"].value_counts()
```

```python
grense = 50
antall = df["league"].value_counts()
små = antall[antall < grense].index
df["league"] = df["league"].replace(små, "annet")

print("Ligaer gruppert som 'annet':", list(små))
df["league"].value_counts()
```

```python
kolonner = [
    "player_id", "name",                                   # for å identifisere spillere, ikke features
    "position", "sub_position", "foot", "league",          # kategorier
    "age", "height_in_cm", "contract_years_left", "contract_missing",
    "games", "minutes", "goals", "assists", "yellow",
    "goals_per90", "assists_per90", "cl_minutes",
    "international_caps", "international_goals",
    "market_value_in_eur",                                 # målvariabel
]
clean = df[kolonner].copy()

print(clean.shape)
print("Manglende verdier totalt:", clean.isna().sum().sum())
```

```python
clean.to_csv("data/model_clean.csv", index=False)
print("Lagret", len(clean), "rader og", clean.shape[1], "kolonner")
```
